```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully.")
```

```python
video_source = "YOUR_ACTUAL_VIDEO_FILENAME.mp4" # Replace with your actual video filename, or 0 for webcam

cap = cv2.VideoCapture(video_source)

if not cap.isOpened():
    print(f"Error: Could not open video source: {video_source}")
else:
    print(f"Video source '{video_source}' opened successfully.")
```

```python
video_source = 0 # Set to 0 for webcam
cap_webcam = cv2.VideoCapture(video_source)

if not cap_webcam.isOpened():
    print(f"Error: Could not open webcam: {video_source}")
else:
    ret, frame_webcam = cap_webcam.read()
    if ret:
        # Convert BGR to RGB for matplotlib display
        frame_rgb = cv2.cvtColor(frame_webcam, cv2.COLOR_BGR2RGB)
        plt.imshow(frame_rgb)
        plt.title("Webcam Frame")
        plt.axis('off') # Hide axes
        plt.show()
        print("Successfully captured and displayed a frame from the webcam.")
    else:
        print("Error: Could not read frame from webcam.")
    cap_webcam.release()

print("Please grant camera permissions to your browser if you encounter an error.")
```

```python
if cap.isOpened():
    ret, frame1 = cap.read()
    if not ret:
        print("Error: Could not read first frame.")
    else:
        # Convert to grayscale for background subtraction and optical flow
        prvs = cv2.cvtColor(frame1, cv2.COLOR_BGR2GRAY)

        # Initialize background subtractor (MOG2 is a common choice)
        fgbg = cv2.createBackgroundSubtractorMOG2(history=500, varThreshold=16, detectShadows=True)

        while True:
            ret, frame2 = cap.read()
            if not ret:
                print("End of video stream.")
                break

            # Convert current frame to grayscale
            next = cv2.cvtColor(frame2, cv2.COLOR_BGR2GRAY)

            # --- 1. Motion Detection (using simple frame differencing for now, will improve later) ---
            # For now, let's use background subtraction as a primary motion indicator
            fgmask = fgbg.apply(frame2)

            # Apply morphological operations to remove noise
            fgmask = cv2.morphologyEx(fgmask, cv2.MORPH_OPEN, kernel=np.ones((3,3),np.uint8))
            fgmask = cv2.morphologyEx(fgmask, cv2.MORPH_CLOSE, kernel=np.ones((5,5),np.uint8))

            # --- 2. Optical Flow (Farneback method) ---
            flow = cv2.calcOpticalFlowFarneback(prvs, next, None, 0.5, 3, 15, 3, 5, 1.2, 0)

            # Visualization of optical flow (optional, for debugging)
            # mag, ang = cv2.cartToPolar(flow[..., 0], flow[..., 1])
            # hsv = np.zeros_like(frame1)
            # hsv[..., 1] = 255
            # hsv[..., 0] = ang * 180 / np.pi / 2
            # hsv[..., 2] = cv2.normalize(mag, None, 0, 255, cv2.NORM_MINMAX)
            # rgb_flow = cv2.cvtColor(hsv, cv2.COLOR_HSV2BGR)

            # --- 3. Object Tracking (placeholder for now, will integrate with detected contours)
            # This will be developed further after motion detection is robust.

            # Combine original frame with motion detection mask for visualization
            # The original frame with contours drawn on it will be more effective.
            display_frame = frame2.copy()
            display_frame[fgmask == 255] = [0, 255, 0] # Highlight detected motion in green

            # Display the processed frame (optional, for real-time visualization in an interactive environment)
            # For Colab, we'll likely save frames or display a final video/summary
            # cv2.imshow('Frame Analysis', display_frame)
            # cv2.imshow('Foreground Mask', fgmask)

            # Press 'q' to exit (for local execution)
            # if cv2.waitKey(1) & 0xFF == ord('q'):
            #     break

            prvs = next # Update previous frame for next iteration

        cap.release()
        cv2.destroyAllWindows()
        print("Video processing complete.")
else:
    print("Video capture not opened. Please check 'video_source' and rerun the previous cell.")
```
